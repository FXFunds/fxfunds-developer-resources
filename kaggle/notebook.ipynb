## Check an address format with the published patterns

```python
def is_valid_address(network, address):
    pattern = nets.set_index('key').loc[network, 'address_pattern']
    return re.fullmatch(pattern.strip('^$'), address) is not None

for net, addr in [('erc20', '0xdAC17F958D2ee523a2206206994597C13D831ec7'), ('trc20', 'TR7NHqjeKQxGTCi8q8ZY4pL8otSzgjLj6t'), ('ton', 'EQCxE6mUtQJKFnGfaROTKOt1lZbDiiX1kCixRv7Nw2Id_sDs'), ('erc20', '0x123')]:
    print(net, addr[:12] + '…', is_valid_address(net, addr))
```
